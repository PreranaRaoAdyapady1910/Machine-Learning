```python
#take weather data of past 1 month of udupi,
#1.find which is the hottest day and coldest day
#2.Calculate the avg temperature of overall month
#3. find which all days has above 35 deg celsius
import pandas as pd
df = pd.read_excel("weather.xlsx")
hottest = df.loc[df["Temperature"].idxmax()]
coldest = df.loc[df["Temperature"].idxmin()]
print("Hottest Day:")
print("Date:", hottest["Date"])
print("Temperature:", hottest["Temperature"], "°C")
print("\nColdest Day:")
print("Date:", coldest["Date"])
print("Temperature:", coldest["Temperature"], "°C")
average_temperature = df["Temperature"].mean()
print("\nAverage Temperature:")
print(round(average_temperature, 2), "°C")
above_35 = df[df["Temperature"] > 35]
print("\nDays above 35°C:")
if above_35.empty:
    print("No days had temperature above 35°C.")
else:
    print(above_35[["Date", "Temperature"]])
```

```python
#assignment-2
#Use titanic dataset to
#1.find total no of passengers
#2.find avg age of survivors and non survivors
#3.list the passengers who are older than 60 years
#4. find the avg fair for each p[assenger class
import pandas as pd
df = pd.read_csv('/content/Titanic-Dataset.csv')
total_passengers = len(df)
print("Total number of passengers are:", total_passengers)
avg_age = df.groupby('Survived')['Age'].mean()
print("\nAverage age of survivors:", avg_age[1])
print("Average age of non-survivors:", avg_age[0])
older_passengers = df[df['Age'] > 60]
print("\nPassengers older than 60 years:")
print(older_passengers)
avg_fare = df.groupby('Pclass')['Fare'].mean()
print("\nAverage fare for each passenger class:")
print(avg_fare)
```
